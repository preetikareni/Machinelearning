```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Ice_cream selling data.csv')
```

```python
df.info()
```

```python
df.corr()
```

```python
x=df[['Temperature (°C)']]
y=df['Ice Cream Sales (units)']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
'''
syntax for linear regression
from sklearn.linear_model import LinearRegression

sklearn-> package/tool
linear model --> module(.py)
linear regression --> class(from linear_model) module
# model creation
model_lr=LinearRegression()
#model training 
fit()
what is fit()?
it learns the pattern from the data(input and output)
model_lr.fit(xtrain,ytrain)# it uses ols
#finds the best fit line (mx+c)
-finds slope and intercept
'''
```

```python
model=LinearRegression()
```

```python
model.fit(xtrain,ytrain)
```

```python
model.intercept_
```

```python
model.coef_ #slope
```

```python
''' what is weights?
It tells the importances of the features(input).
Based on weights model decides the importance of features(input)

'''
```

Actual data vs model predicted data

```python
ytrain_predicted=model.predict(xtrain)
```

```python
plt.scatter(xtrain,ytrain) # actual input and actual output
plt.scatter(xtrain,ytrain_predicted)# actual input and predicted output
```

Conclusion:
  Model is not able to capture the patterns.

Reason:
 The data is non-linear and LinearRegression can capture only linear data.

 

```python
''' 
why we're comparing this?
1)plt.scatter(xtrain,ytrain)#actual input vs actual output
2)plt.scatter(xtrain,ytrain_predicted)#actual input vs predicted output

answer:
From this comparsion between actual pattern and predicted pattern
we can coclude how well the model captures the pattern during model training'''
```

```python
ytest_predicted=model.predict(xtest)
```

```python
plt.scatter(xtest,ytest)
plt.scatter(xtest,ytest_predicted)
```

Conclusion:
  Model is not able to capture the test patterns.

Reason:
 The data is non-linear and LinearRegression can capture only linear data.


```python
train_data = pd.DataFrame({
    'xtrain':xtrain['Temperature (°C)'],
    'ytrain':ytrain,
    'ytrain_pred':ytrain_predicted
})
train_data['Error']=train_data['ytrain']-train_data['ytrain_pred']
```
